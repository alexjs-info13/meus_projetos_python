```python
from IPython.display import clear_output

def exibir_tabuleiro(tabuleiro):
    """Imprime o tabuleiro com cores e organização."""
    # Cores organizadas: \033[1;34m é Azul, \033[1;31m é Vermelho, \033[0m reseta a cor
    print("\n")
    print(f"  {tabuleiro[0]}  |  {tabuleiro[1]}  |  {tabuleiro[2]} ")
    print("-----|-----|-----")
    print(f"  {tabuleiro[3]}  |  {tabuleiro[4]}  |  {tabuleiro[5]} ")
    print("-----|-----|-----")
    print(f"  {tabuleiro[6]}  |  {tabuleiro[7]}  |  {tabuleiro[8]} ")
    print("\n")

def verificar_vitoria(tabuleiro, jogador):
    condicoes = [
        [0, 1, 2], [3, 4, 5], [6, 7, 8],  # Linhas
        [0, 3, 6], [1, 4, 7], [2, 5, 8],  # Colunas
        [0, 4, 8], [2, 4, 6]              # Diagonais
    ]
    for condicao in condicoes:
        if tabuleiro[condicao[0]] == tabuleiro[condicao[1]] == tabuleiro[condicao[2]] == jogador:
            return True
    return False

def jogo_da_velha():
    tabuleiro = [str(i) for i in range(1, 10)]
    jogador_atual = "\033[1;34mX\033[0m"  # X será Azul
    jogadas = 0

    while jogadas < 9:
        clear_output() # 🌟 Isso limpa a tela no Colab a cada jogada!
        print("=" * 35)
        print("      🎮 JOGO DA VELHA 🎮      ")
        print("=" * 35)

        exibir_tabuleiro(tabuleiro)

        try:
            escolha = int(input(f"Jogador [{jogador_atual}], escolha uma posição (1-9): ")) - 1
        except ValueError:
            print("❌ Por favor, digite um número válido!")
            input("Aperte Enter para continuar...")
            continue

        if escolha < 0 or escolha > 8 or tabuleiro[escolha] in ["\033[1;34mX\033[0m", "\033[1;31mO\033[0m"]:
            print("⚠️ Posição inválida ou já ocupada!")
            input("Aperte Enter para continuar...")
            continue

        tabuleiro[escolha] = jogador_atual
        jogadas += 1

        # Verifica vitória usando a string colorida do jogador atual
        if verificar_vitoria(tabuleiro, jogador_atual):
            clear_output()
            exibir_tabuleiro(tabuleiro)
            print(f"🎉 PARABÉNS! O jogador [{jogador_atual}] venceu o jogo! 🏆\n")
            return

        # Alterna o jogador e a cor (X é Azul, O é Vermelho)
        jogador_atual = "\033[1;31mO\033[0m" if jogador_atual == "\033[1;34mX\033[0m" else "\033[1;34mX\033[0m"

    clear_output()
    exibir_tabuleiro(tabuleiro)
    print("👵 O jogo empatou! Deu velha!")

# Executa o jogo
jogo_da_velha()
```
